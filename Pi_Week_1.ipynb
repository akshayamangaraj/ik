# Two Sum


```python
# Two SUm
arr = [2,7,11,15]
target = 9
for i in range(len(arr)):
    for j in range(i+1,len(arr)):
        if arr[i]+arr[j]==target:
            print("i={0} j={1}",i,j)
            # Format the message and print
            print("The indices are: {} and {}".format(i, j))
```
